```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Price": [20, 25, 30, 35, 40],
    "Bottles Sold": [120, 110, 90, 80, 70]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df["Revenue"] = df["Price"] * df["Bottles Sold"]
```

```python
print(df)
```

```python
highest_revenue = df["Revenue"].max()
print("Highest Revenue:", highest_revenue)
```

```python
best_price = df.loc[df["Revenue"].idxmax(), "Price"]
print("Price that generated the highest revenue:", best_price)
```

```python
plt.figure(figsize=(6,4))
plt.bar(df["Price"], df["Revenue"])
plt.title("Price vs Revenue")
plt.xlabel("Price (₹)")
plt.ylabel("Revenue (₹)")
plt.show()
```

Conclusion

The analysis shows that the highest revenue is ₹2800, achieved at both ₹35 and ₹40.
Since ₹35 sells more bottles than ₹40, the shop owner should choose ₹35.
This price generates maximum revenue while maintaining higher sales volume.
